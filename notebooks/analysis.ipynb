```python
from climbing_performance.models import Rider, Bike, Climb
from climbing_performance.metrics import summarise_full_performance

rider = Rider.default()
bike = Bike.default()
climb = Climb.default()

summary = summarise_full_performance(rider, bike, climb)
summary
```

```python
climb_sea = Climb.default()
climb_high = Climb.default()
climb_high.avg_altitude_m = 2000.0

sea = summarise_full_performance(rider, bike, climb_sea)
high = summarise_full_performance(rider, bike, climb_high)

print("Sea level W/kg:", sea["watts_per_kg"])
print("Sea level aSLP W/kg:", sea["aslp_w_per_kg"])
print("High altitude W/kg:", high["watts_per_kg"])
print("High altitude aSLP W/kg:", high["aslp_w_per_kg"])
print("High altitude loss estimate:", high["altitude_power_loss_percent"])
```

```python
climb_elite = Climb(
    distance_m=13800.0,
    elevation_gain_m=1100.0,
    time_s=2400.0,
    avg_altitude_m=1400.0,
)

elite = summarise_full_performance(rider, bike, climb_elite)
elite
```
